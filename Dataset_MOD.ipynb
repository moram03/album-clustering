## Modify / Reshape Dataset

```python
import os
import shutil

def move_and_rename_subfolders(main_folder):
    # Iterate through each folder in the main folder
    for folder_event_name in os.listdir(main_folder):
        folder_event_path = os.path.join(main_folder, folder_event_name)
        
        # Make sure it's a directory
        if os.path.isdir(folder_event_path):
            # Iterate through subfolders
            for subfolder_name in os.listdir(folder_event_path):
                subfolder_path = os.path.join(folder_event_path, subfolder_name)
                
                if os.path.isdir(subfolder_path):
                    # Create new folder name
                    new_name = f"{folder_event_name}"#_{subfolder_name}"
                    new_path = os.path.join(main_folder, new_name)

                    print(new_name)
                    
                    # Move and rename the subfolder
                    shutil.move(subfolder_path, new_path)
                    print(f"Moved: {subfolder_path} --> {new_path}")

            # Optionally, remove the now empty parent folder
            if not os.listdir(folder_event_path):  # Folder is empty
                os.rmdir(folder_event_path)
                print(f"Removed empty folder: {folder_event_path}")

# Example usage:
# Replace this with the actual path to your main folder
main_folder_path = "./pec/images"
move_and_rename_subfolders(main_folder_path)
```

## Optimal Size + Deleting Non-Image Files

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from collections import Counter

def analyze_image_sizes(root_folder):
    """
    Analyzes all images in the given folder and determines the best resizing size
    based on the median dimensions of all images.

    Parameters:
        root_folder (str): The path to the root folder containing images.

    Returns:
        tuple: (best_width, best_height)
    """
    widths = []
    heights = []
    
    # Scan all subfolders
    for subfolder in os.listdir(root_folder):
        subfolder_path = os.path.join(root_folder, subfolder)
        if os.path.isdir(subfolder_path):
            for file in os.listdir(subfolder_path):
                file_path = os.path.join(subfolder_path, file)
                try:
                    with Image.open(file_path) as img:
                        width, height = img.size
                        widths.append(width)
                        heights.append(height)
                except Exception as e:
                    #print(f"Error processing {file_path}: {e}")
                    os.remove(file_path)  # <-- This line deletes non-image files

    if not widths or not heights:
        print("No images found!")
        return None

    # Compute the best resizing size (median dimensions)
    best_width = int(np.median(widths))
    best_height = int(np.median(heights))

    print(f"Best resizing dimensions: {best_width}x{best_height}")

    # Visualize the distribution of image sizes
    visualize_image_sizes(widths, heights)

    return best_width, best_height


def visualize_image_sizes(widths, heights):
    """
    Creates histograms for image widths and heights to visualize size distribution.

    Parameters:
        widths (list): List of image widths.
        heights (list): List of image heights.
    """
    plt.figure(figsize=(12, 5))

    # Plot Widths
    plt.subplot(1, 2, 1)
    plt.hist(widths, bins=20, color="blue", alpha=0.7, edgecolor="black")
    plt.xlabel("Width (pixels)")
    plt.ylabel("Frequency")
    plt.title("Image Width Distribution")

    # Plot Heights
    plt.subplot(1, 2, 2)
    plt.hist(heights, bins=20, color="green", alpha=0.7, edgecolor="black")
    plt.xlabel("Height (pixels)")
    plt.ylabel("Frequency")
    plt.title("Image Height Distribution")

    plt.tight_layout()
    plt.show()


# Example usage
main_folder_path = "./pec/images"
best_size = analyze_image_sizes(main_folder_path)
```

```python
import os
import shutil

def move_images_to_parent_subfolder(base_directory):
    # List of image extensions to move
    image_extensions = ('.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp')

    # Walk through all subdirectories and files in the base directory
    for dirpath, dirnames, filenames in os.walk(base_directory):
        # Only process sub-subfolders (not the base folder or first-level subfolders)
        if dirpath != base_directory:
            parent_subfolder = os.path.basename(dirpath)

            # Skip the parent folder itself and just look at the subfolders inside it
            if parent_subfolder != base_directory:
                for filename in filenames:
                    file_path = os.path.join(dirpath, filename)
                    
                    # Check if the file is an image
                    if filename.lower().endswith(image_extensions):
                        # Define the destination path in the parent subfolder (move to parent folder)
                        parent_folder = os.path.dirname(dirpath)  # Parent of the current subfolder
                        destination = os.path.join(parent_folder, filename)

                        # Avoid overwriting files by renaming if necessary
                        if os.path.exists(destination):
                            name, ext = os.path.splitext(filename)
                            counter = 1
                            new_filename = f"{name}_{counter}{ext}"
                            destination = os.path.join(parent_folder, new_filename)
                            while os.path.exists(destination):
                                counter += 1
                                new_filename = f"{name}_{counter}{ext}"
                                destination = os.path.join(parent_folder, new_filename)

                        # Move the image file to the parent subfolder
                        shutil.move(file_path, destination)
                        print(f"Moved: {filename} from {dirpath} to {parent_folder}")

# Example usage:
base_directory = './pec/images/wedding'  # Change this to your target directory
move_images_to_parent_subfolder(base_directory)
```
